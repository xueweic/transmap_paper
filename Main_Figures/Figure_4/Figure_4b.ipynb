# Figure 4b. Cell types in *cis* and *trans*

Percentage of the 4,023 *cis*-*trans* colocalized loci whose *cis*-eQTL (initiating side) or *trans*-eQTL (recipient side, eGene union eProgram) involves at least one neuronal (Exc, Inh) or glial (Mic, Ast, Oli, OPC) cell type. A locus can count in both groups, so the two groups do not sum to 100%. Descriptive percentages; no statistical test.

Input: `data/Figure_4b_neuron_glia_pct.rds`

```r
library(ggplot2)
res <- readRDS("data/Figure_4b_neuron_glia_pct.rds")
res
```

```r
p <- ggplot(res, aes(x = group, y = pct, fill = qtl_type)) +
  geom_col(position = position_dodge(width = 0.70), width = 0.60,
           colour = "white", linewidth = 0.45) +
  geom_text(aes(label = sprintf("%.1f%%", pct)),
            position = position_dodge(width = 0.70), vjust = -0.4, size = 5) +
  scale_fill_manual(values = c("cis-eQTL" = "#D4A72C", "trans-eQTL" = "#58769D"),
                    labels = c(expression(italic(cis) * "-eQTL"),
                               expression(italic(trans) * "-eQTL")),
                    name = NULL) +
  scale_y_continuous(expand = expansion(mult = c(0, 0.16))) +
  labs(title = "% of colocalized loci", x = NULL, y = NULL) +
  theme_classic(base_size = 18) +
  theme(plot.title = element_text(hjust = 0.5),
        legend.position = "inside", legend.position.inside = c(0.05, 0.95),
        legend.justification = c(0, 1),
        panel.grid.major.y = element_line(colour = "grey91", linewidth = 0.35))

options(repr.plot.width = 5.5, repr.plot.height = 4.5)
p
```
