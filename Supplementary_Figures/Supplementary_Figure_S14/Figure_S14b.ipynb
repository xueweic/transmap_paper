# Supplementary Figure S14b. Colocalized trans phenotypes by recipient context at the MS4A locus

Number of *trans*-eGenes plus *trans*-ePrograms that colocalize with AD in the chr11 MS4A-family colocalization set CoS3 (lead variant chr11:60171834:T:G; ColocBoost, `get_robust_colocalization(pvalue_cutoff = 1e-4)`), by the cell type / tissue in which the *trans* phenotype was measured. Protein and glycosylation *trans*-biomarker QTLs in the same set are not counted. The counts are computed from the ColocBoost result (the author's notebook hard-coded them; the values agree). Inhibitory neurons (9 *trans*-eGenes and 3 *trans*-eProgram components: module 59 and two WPCs of module 115) are highlighted.

Input: `data/Figure_S14b_MS4A_trans_context_counts.rds`


```r
library(ggplot2)
dat <- readRDS("data/Figure_S14b_MS4A_trans_context_counts.rds")
dat$counts
subset(dat$outcomes, type %in% c("trans-eGene", "trans-eProgram"))[, c("context", "type", "gene", "program")]
```

```r
d <- dat$counts
d$context <- factor(d$context, levels = rev(d$context))
p <- ggplot(d, aes(x = total, y = context, fill = context == "Inh")) +
  geom_col(width = 0.68) +
  geom_text(aes(label = total), hjust = -0.25, size = 5) +
  scale_fill_manual(values = c(`FALSE` = "#9CB2BE", `TRUE` = "#B35A27"), guide = "none") +
  scale_x_continuous(limits = c(0, 14), breaks = seq(0, 12, 2), expand = c(0, 0)) +
  labs(title = "Trans outcomes by recipient context", x = "Number of trans outcomes", y = NULL) +
  theme_classic(base_size = 14) +
  theme(plot.title = element_text(hjust = 0.5))
options(repr.plot.width = 8, repr.plot.height = 3.5)
p
```
