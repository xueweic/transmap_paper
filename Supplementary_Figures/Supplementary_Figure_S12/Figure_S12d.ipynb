# Supplementary Figure S12d. Cell-type involvement of AD loci, splitting *trans*-eGene and *trans*-eProgram

As in S12c (112-event version), but the *trans* evidence is split into *trans*-eGene (denominator: 107 events with any *trans*-eGene colocalization) and *trans*-eProgram (local program; denominator: 103 events); the *cis*-eQTL bars (denominator 65) are identical to S12c. Percentages are n_loci / total x 100; counts only.

Input: `data/Figure_S12d_celltype_pct_split.rds`.

```r
suppressPackageStartupMessages(library(ggplot2))
d <- readRDS("data/Figure_S12d_celltype_pct_split.rds")
d
```

```r
cat_colors <- c("cis-eQTL" = "#2171B5", "trans-eGene" = "#E69F00", "trans-eProgram" = "#CC3311")
cat_labels <- c(expression(italic(cis) * "-eQTL"), expression(italic(trans) * "-eGene"),
                expression(italic(trans) * "-eProgram"))
p <- ggplot(d, aes(cell_type, pct, fill = category)) +
  geom_col(position = position_dodge(width = 0.78), width = 0.7, color = "black", linewidth = 0.25, alpha = 0.9) +
  geom_text(aes(label = ifelse(n_loci > 0, sprintf("%.0f%%", pct), "")),
            position = position_dodge(width = 0.78), vjust = -0.35, size = 3.3) +
  scale_fill_manual(values = cat_colors, labels = cat_labels, name = NULL) +
  scale_y_continuous(expand = expansion(mult = c(0, 0.15))) +
  labs(x = NULL, y = "% of AD colocalized loci", title = "Cell-type involvement in AD colocalization") +
  theme_classic(base_size = 13) +
  theme(plot.title = element_text(hjust = 0.5), legend.position = "bottom",
        panel.grid.major.y = element_line(color = "grey91", linewidth = 0.35))

options(repr.plot.width = 8, repr.plot.height = 4.2)
p
```
