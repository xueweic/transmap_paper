# Supplementary Figure S12c. Cell-type involvement of AD loci via *cis*- or *trans*-eQTL

Percentage of AD colocalization events (112-event version, 228 CoS) that involve each snRNA-seq cell type through a *cis*-eQTL (denominator: 65 events with any *cis*-eGene colocalization in any context) or a *trans*-eQTL (*trans*-eGene, *trans*-eProgram or *trans* global program; denominator: 111 events with any *trans* colocalization in any context). Percentages are n_loci / total x 100; counts only, no statistical test.

Input: `data/Figure_S12c_celltype_pct_cis_trans.rds`.

```r
suppressPackageStartupMessages(library(ggplot2))
d <- readRDS("data/Figure_S12c_celltype_pct_cis_trans.rds")
d
```

```r
cat_colors <- c("cis-eQTL" = "#2171B5", "trans-eQTL" = "#D55E00")
cat_labels <- c(expression(italic(cis) * "-eQTL"), expression(italic(trans) * "-eQTL"))
p <- ggplot(d, aes(cell_type, pct, fill = category)) +
  geom_col(position = position_dodge(width = 0.78), width = 0.7, color = "black", linewidth = 0.25, alpha = 0.9) +
  geom_text(aes(label = ifelse(n_loci > 0, sprintf("%.0f%%", pct), "")),
            position = position_dodge(width = 0.78), vjust = -0.35, size = 3.3) +
  scale_fill_manual(values = cat_colors, labels = cat_labels, name = NULL) +
  scale_y_continuous(expand = expansion(mult = c(0, 0.15))) +
  labs(x = NULL, y = "% of AD colocalized loci", title = "Cell-type involvement in AD colocalization") +
  theme_classic(base_size = 13) +
  theme(plot.title = element_text(hjust = 0.5), legend.position = "bottom",
        panel.grid.major.y = element_line(color = "grey91", linewidth = 0.35))

options(repr.plot.width = 7, repr.plot.height = 4.2)
p
```
