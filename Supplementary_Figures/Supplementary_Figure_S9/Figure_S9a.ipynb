# Supplementary Figure S9a. Cell-type involvement in *cis* and *trans*

Percentage of the 4,023 *cis*-*trans* colocalized loci (CoS supported by a *cis*-eQTL and by at least one *trans*-eGene or *trans*-eProgram) whose *cis*-eQTL (DeJager snRNA-seq) or *trans*-eQTL (*trans*-eGene union *trans*-eProgram) signal involves each of the six brain cell types. A locus can involve several cell types, so percentages within a QTL type do not sum to 100%; loci supported only by bulk (ROSMAP DLPFC/AC/PCC) count in the denominator but in no cell type. Descriptive percentages; no statistical test.

Input: `data/Figure_S9a_celltype_involvement.rds`

```r
library(ggplot2)
res <- readRDS("data/Figure_S9a_celltype_involvement.rds")
res
```

```r
p <- ggplot(res, aes(x = cell_type, y = pct, fill = qtl_type)) +
  geom_col(position = position_dodge(width = 0.74), width = 0.64,
           colour = "white", linewidth = 0.45) +
  geom_text(aes(label = sprintf("%.0f%%", pct)),
            position = position_dodge(width = 0.74), vjust = -0.35, size = 4.5) +
  scale_fill_manual(values = c("cis-eQTL" = "#D4A72C", "trans-eQTL" = "#58769D"),
                    labels = c(expression(italic(cis) * "-eQTL"),
                               expression(italic(trans) * "-eQTL")),
                    name = NULL) +
  scale_y_continuous(expand = expansion(mult = c(0, 0.25))) +
  labs(x = NULL, y = "% of colocalized loci",
       title = "Cell-type involvement in cis- and trans-eQTL colocalization") +
  theme_classic(base_size = 16) +
  theme(plot.title = element_text(hjust = 0.5, size = 16),
        legend.position = "inside", legend.position.inside = c(0.98, 0.98),
        legend.justification = c(1, 1), legend.direction = "horizontal",
        panel.grid.major.y = element_line(colour = "grey91", linewidth = 0.35))

options(repr.plot.width = 8, repr.plot.height = 4.5)
p
```
