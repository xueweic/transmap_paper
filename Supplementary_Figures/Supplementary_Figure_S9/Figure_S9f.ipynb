# Supplementary Figure S9f. Excitatory-neuron credible sets after subsampling

Number of excitatory-neuron *cis*-eGene credible sets (rows of the SuSiE credible-set tables) and *trans*-eGene credible sets (unique TransMap fine-mapping identifiers with a 95% credible set) using all excitatory-neuron nuclei (Full) or random subsets of 100,000, 200,000 and 300,000 nuclei. `pct_of_full` gives each count relative to the full data. Descriptive counts; no statistical test.

Input: `data/Figure_S9f_subsample_credible_sets.rds`

```r
library(ggplot2)
library(scales)
res <- readRDS("data/Figure_S9f_subsample_credible_sets.rds")
res
```

```r
cols <- c("Full" = "#1F2D4B", "100K" = "#3F5B7A", "200K" = "#7A9AAF", "300K" = "#B5C6D2")
lab <- c(cis = "cis-eGene credible sets (Exc)", trans = "trans-eGene credible sets (Exc)")

p <- ggplot(res, aes(x = sample, y = n_cs, fill = sample)) +
  geom_col(width = 0.68, colour = "grey20", linewidth = 0.35) +
  geom_text(aes(label = comma(n_cs)), vjust = -0.45, size = 4.2) +
  facet_wrap(~ layer, scales = "free_y", labeller = as_labeller(lab)) +
  scale_fill_manual(values = cols, guide = "none") +
  scale_y_continuous(labels = comma, expand = expansion(mult = c(0, 0.12))) +
  labs(x = "Number of nuclei", y = "# credible sets") +
  theme_classic(base_size = 15) +
  theme(strip.background = element_blank(), strip.text = element_text(size = 14),
        panel.grid.major.y = element_line(colour = "grey92", linewidth = 0.3))

options(repr.plot.width = 10, repr.plot.height = 4.2)
p
```
