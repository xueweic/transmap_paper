# Supplementary Figure S7d. Downstream *trans*-regulatory layers of *cis*-eQTL loci

For each of the 15,435 *cis*-eQTL colocalized loci, the number of downstream *trans* layers it also colocalizes with (0-3; layers: *trans*-eGene, *trans*-eProgram, and *trans*-biomarker = union of protein-, glycoprotein- and metabolite-QTL). Bars show the proportion of *cis*-eQTL loci in each class. Loci were first filtered to keep *cis*-eQTL loci and *trans* loci that are also biomarker-QTL loci (as in main Figure 3a); the filter does not remove any *cis*-eQTL locus. No statistical test is involved.

Input: `data/Figure_S7d_cis_trans_layers.rds`.

```r
library(ggplot2)
s7d <- readRDS("data/Figure_S7d_cis_trans_layers.rds")
s7d
```

```r
s7d$n_trans_layers <- factor(s7d$n_trans_layers, levels = 0:3)
s7d$label <- sprintf("%.1f%%", 100 * s7d$proportion)
pal <- c("0" = "#B8C7D9", "1" = "#5BA3A6", "2" = "#7A6BB7", "3" = "#C95F3E")

p <- ggplot(s7d, aes(x = n_trans_layers, y = proportion, fill = n_trans_layers)) +
  geom_col(width = 0.62, color = "white") +
  geom_text(aes(label = label), vjust = -0.5, size = 5, color = "grey15") +
  scale_fill_manual(values = pal, name = NULL) +
  scale_y_continuous(breaks = seq(0, 1, 0.25), limits = c(0, 0.85), expand = expansion(mult = c(0, 0.05))) +
  labs(x = expression(paste("Number of downstream ", italic("trans"), "-effects")), y = "Proportion") +
  guides(fill = guide_legend(nrow = 1)) +
  theme_classic(base_size = 15) +
  theme(legend.position = c(0.7, 0.9), legend.justification = c(0.5, 1))

options(repr.plot.width = 8, repr.plot.height = 3.5)
p
```
